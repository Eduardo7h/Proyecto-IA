```python
import numpy as np 
import pandas as pd 

#leer datos de un archivo csv
df = pd.read_csv('data_studentlist.csv',header = 'infer')
#pasamos la altura de cm a m  
df['Height'] = df['Height'].apply(lambda x : x/100)

#promedio de la altura de los varones 
HM = df.loc[(df['Gender'] =='M'),'Height'].mean().round(2)

#promedio de la altura de las estudiantes
HF = df.loc[(df['Gender'] == 'F'),'Height'].mean().round(2)  

#promedio del peso de los hombres 
WM = df.loc[(df['Gender'] == 'M'),'Weight'].mean().round(2)

#promedio de la altura de las estudiantes
WF = df.loc[(df['Gender'] == 'F'),'Weight'].mean().round(2)  

#altura maxima entre los varones 
HM_max = df.loc[(df['Gender'] == 'M' ),'Height'].max()
HM_max = round(HM_max,2)

#altura minima entrea las mujeres 
HF_min = df.loc[(df['Gender']=='F'),'Height'].min()
HF_min = round(HF_min,2)

#peso mas ligero entre los varones mayores a 1.75
WM1_min = df.loc[(df['Gender'] == 'M') & (df['Height']>1.75), 'Weight'].min()

#mayor peso etre las mujeres menores a 1.60m 
WF1_max = df.loc[(df['Gender'] == 'F') & (df['Height'] < 1.60),'Weight'].max()

# La nota promedio de los alumnos (hombres y mujeres) sin ausencias ('N')
NP = df.loc[(df['Absence'] == 'N'),'Grade'].mean().round(2)

# La nota promedio de los alumnos (hombres y mujeres) con ausencias ('Y')
YP = df.loc[(df['Absence'] == 'Y'),'Grade'].mean().round(2)

# Altura promedio de los alumnos (hombres y mujeres) con grupo sanguíneo 'A' o 'AB'
ABP = df.loc[(df['Bloodtype'] == 'A') | (df['Bloodtype'] == 'AB'),'Height'].mean().round(2)

#Altura promedio de los alumnos varones con grupo sanguíneo 'A' o 'AB'
ABPM = df.loc[(df['Gender']== 'M') & ((df['Bloodtype'] == 'A') | (df['Bloodtype']  == 'AB')),'Height'].mean().round(2)

#Edad promedio de los alumnos (hombres y mujeres) con absentismo ('Y') cuya nota es mayor o igual a 3
AP = df.loc[((df['Absence'] == 'Y') & (df['Grade'] >= 3)),'Age'].mean().round()



print('El promedio de la altura de los varones es -> ',HM,'m')
print('El promedio de la altura de las mujeres es -> ',HF,'m')
print('El promedio del peso de los hombres es -> ',WM,'Kg')
print('El promedio del peso de las mujeres es -> ',WF,'kg')
print('la altura maxima de los varoes es ->',HM_max,'m')
print ('La aultra minima entre las mujeres es ->',HF_min,'m')
print('El peso mas liguereo entre los varones mayotes a 1.75 m es ->',WM1_min,'Kg')
print('el mayor peso entre las mujeres menore a 1.60m es ->',WF1_max,'Kg')
print('La nota promedio de los alumnos (hombres y mujeres) sin ausencias (N) es ->',NP)
print('La nota promedio de los alumnos (hombres y mujeres) con ausencias (Y) es ->',YP)
print('Altura promedio de los alumnos (hombres y mujeres) con grupo sanguíneo A o AB es ->',ABP,'m')
print('Altura promedio de los alumnos varones con grupo sanguíneo A o AB -> ', ABPM,'m')
print('Edad promedio de los alumnos (hombres y mujeres) con absentismo (Y) cuya nota es mayor o igual a 3 ->',AP,'años')


```

```python
import numpy as np 
import pandas as pd 

#leer datos de una archivo csv 
df1 = pd.read_csv('data_census.csv',header = 'infer')

#modfica el tipo de dato de la columna 'Population'
df1['Population'] = df1['Population'].apply(lambda x: x.replace(',','')).astype(int) 

#Suma la población con ProvinceCode igual a 115 o 116
PT = df1.loc[((df1["Province Code"] == 115 )| (df1["Province Code"] == 116)),'Population'].sum()
print('Población total con ProvinceCode igual a 115 o 116 es ->',PT,'\n')

#Población promedio de las ciudades donde hay más hombres (GenderRatio > 1)
PP = df1.loc[(df1['Gender Ratio']> 1),'Population'].mean().round(2)
print('Población promedio de las ciudades donde hay más hombres (GenderRatio > 1) es ->',PP,'\n')

#Lugares con más hombres (GenderRatio > 1) y menos de 2 personas por hogar
LP = df1.loc[(df1['Gender Ratio']>1)& (df1['Person\'s Household']<2)]
print('Lugares con más hombres (GenderRatio > 1) y menos de 2 personas por hogar\n',LP,'\n')

#Ordena el DataFrame en orden ascendente de 'Households' y muestra los 10 primeros
df1.sort_values(by='Households', ascending=True, inplace=True)
print('DataFrame ordenado en orden ascendente deHouseholds:\n')
df1.head(10)
```

```python
import pandas as pd
import numpy as np 

#leer datos de un archivo csv
df2 = pd.read_csv('data_sales.csv', header='infer')

#Añade una nueva variable Amount = UnitPrice x Units y muestra las 5 primeras observaciones
df2['Amount'] = df2['Unit Price'] * df2['Units']
print("Tabla actualizada: \n",df2.head(5))

#Calcula el precio unitario promedio de cada región. Utiliza el método groupby()
Pr = df2.groupby('Region')['Unit Price'].mean().round(2)
print('\nPromedio de Unit Price usando groupby: \n',Pr)

#Calcula el precio unitario promedio de cada región. Utiliza el método pivot_table()
dfr = pd.pivot_table(df2,index ='Region'  ,values ='Unit Price', aggfunc='mean').round(2)
print('\nPromedio de Unit Price usando pivot_table: \n',dfr.head())

'''
Calcula el total de unidades para cada región y tipo de artículo en una línea de código
Utiliza el método pivot_table() y rellena los valores faltantes con 0.
'''
dfr2 = pd.pivot_table(df2,index=['Region','Item'], values = 'Units' ,aggfunc='sum' ,fill_value = 0 )
print('\nTotal de unidades para cada region:\n',dfr2)

'''
Calcula el Total de ventas para cada región y el tipo de artículo en una línea de código
Utiliza el método pivot_table() y rellena los valores faltantes con 0
'''
dfr3=pd.pivot_table(df2,index=['Region','Item'], values = 'Amount',aggfunc='sum' , fill_value=0)
print('\nTotal de ventas para cada reigion y tipo de articilo:\n')
dfr3.head(15)
```
