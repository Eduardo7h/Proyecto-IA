```python
import pandas as pd 
import numpy as np 
```

```python
mis_indice = ['a','b','c']
mis_valores = [222,333,444]
mi_diccionario ={'a':222,'b':333,'c':444}

#Crear una serie con pandas 
pd.Series(mis_valores)

#index - indices de la serie 
pd.Series(data = mis_valores, index  = mis_indice)
```

```python
# para los diccionarios la llave seran los indices 
# de la serie 
pd.Series(mi_diccionario)
```

```python
s = pd.Series(data = [111,222,333,444],
                index = ['a','b','c','d'])
# index - indices
s.index 
# dttypes - tipos de datos en este caso int 
s.dtypes
# values - valores
s.values

# acceder a un elemento de la serie
s[1]
s['a']
s.loc['a']

# para acceder a mas de un elemento  
s[['a','b']]
```

```python
S1 = pd.Series(data = [1,2,3,4],
               index = ['d','b','c','a'])
S2 = pd.Series(data = [1,2,3,4],
               index = ['a','b','d','e'])

# suma entrada a entrada tamando en cuenta la existencia de los datos 
S1 + S2

#Promedio de los valores de la serie 
S1.mean()
#mediana 
S1.median()
#indice maximo
S1.max()
#desviacion estandar 
S1.std()
#ordena la serie apartir de los valores 
S1.sort_values()
#ordena la serie aparir de los indices 
S1.sort_index()
```
