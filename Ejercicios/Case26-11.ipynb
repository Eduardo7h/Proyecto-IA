```python
import matplotlib.pyplot as plt
import numpy as np 
```

```python
x = np.random.rand(10000)*10+3
y = np.random.rand(10000)*10+5
z = np.random.rand(10000)*10+1

plt.boxplot([x,y,z])
plt.title('Box plot')
plt.show()
```

```python
#orientacion del grafico
plt.boxplot([x,y,z],vert=False)
plt.title('Box plot')
plt.show()

#rellenado de color en cajas 
plt.boxplot([x,y,z],patch_artist=True,
            boxprops=dict(facecolor = 'blue', color = 'red',linewidth = 2,linestyle='--'),
            medianprops=dict(color = 'orange',linewidth = 2),
            whiskerprops=dict(color = "purple",linewidth = 2))
plt.title('Box plot')
plt.xticks([1,2,3],['x','y','z'])
plt.show()
```

```python
#rellenado de color en cajas 
plt.boxplot([x,y,z],patch_artist=True,
            boxprops=dict(facecolor = 'blue', color = 'red',linewidth = 2,linestyle='--'),
            medianprops=dict(color = 'orange',linewidth = 2),
            whiskerprops=dict(color = "purple",linewidth = 2),
            flierprops=dict(marker='^',
                            markerfacecolor = 'red',
                            markersize = 6))
plt.title('Box plot')
plt.xticks([1,2,3],['x','y','z'])
plt.show()
```

```python
#arreglo de numeros que devuelve un arreglo de 100 elementos con numeros entre o y 100
x = np.linspace(0,10,100)
y=np.sin(x)
g =np.cos(x)

plt.plot(x,y,color = 'blue',linewidth=2, linestyle = ':',label ='sine wave')
plt.plot(x,g,color = 'red',linewidth=2, linestyle = ':',label ='cos wave')
plt.title("line plot")
plt.xlabel('x')
plt.ylabel('sin')
#plt.xlim([0,5])
#plt.ylim([-1,1])
plt.legend()
plt.show()
```
