```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns

from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs 
```

```python
X1 ,label1 = make_blobs(n_samples = 100, n_features = 2, centers = 2,cluster_std = 5, random_state = 123 ) ## crea nubes de datos(puntos)
plt.scatter(X1[:,0],X1[:,1], c = label1 , alpha=0.7)
plt.title("Conjuntos de datos 1")
plt.show()
```

```python
X2,label2 = make_blobs(n_samples=100, n_features=2, centers=3, cluster_std=1,random_state=321)
plt.scatter(X2[:,0],X2[:,1], c = label2 , alpha=0.7) 
plt.title("Conjuntos de datos 2")
plt.show()
```

```python
Kmasn1 = KMeans(n_clusters=2 , random_state=123,n_init='auto' ) #'''n_cluster numeros de grupos, n_init -> indica que tanto se va a ejecutar el algoritmo com fiferentes inicializaciones de los centroides  '''
Kmasn1.fit(X1)
```

```python
## Centroides 
Kmasn1.cluster_centers_
```

```python
## creacion de la tabla de dispersion 
colors = {0:'red',1:'purple',2:'green'}
plt.scatter(X1[:,0],X1[:,1], c= pd.Series(Kmasn1.labels_).apply(lambda x: colors[x]),alpha=0.7)
plt.title('K-Means en conjunto de datos 1')
plt.show()
```

```python
Kmasn2 = KMeans(n_clusters=3 , random_state=123,n_init='auto' ) #'''n_cluster numeros de grupos, n_init -> indica que tanto se va a ejecutar el algoritmo com fiferentes inicializaciones de los centroides  '''
Kmasn2.fit(X1)

## creacion de la tabla de dispersion 
colors = {0:'blue',1:'red',2:'green'}
plt.scatter(X1[:,0],X1[:,1], c= pd.Series(Kmasn2.labels_).apply(lambda x: colors[x]),alpha=0.7)
plt.title('K-Means en conjunto de datos 2')
plt.show()
```

```python
Kmasn3 = KMeans(n_clusters=2 , random_state=123,n_init='auto' ) #'''n_cluster numeros de grupos, n_init -> indica que tanto se va a ejecutar el algoritmo con fiferentes inicializaciones de los centroides  '''
Kmasn3.fit(X2)

## creacion de la tabla de dispersion 
colors = {0:'blue',1:'red',2:'green'}
plt.scatter(X2[:,0],X2[:,1], c= pd.Series(Kmasn3.labels_).apply(lambda x: colors[x]),alpha=0.7)
plt.title('K-Means en conjunto de datos 2')
plt.show()
```

```python
df = sns.load_dataset('iris')
X = df.drop(columns=['species'])
Y = df['species']
header_x = X.columns 
X.head()
```

```python
kmeans = KMeans(n_clusters=3 , random_state=123,n_init='auto')
kmeans.fit(X)
res = pd.Series(kmeans.labels_)
```

```python
# especies en funcion de cada grupo 
# mascarass booleanas que quedan en funcion de los grupos 

case0 = Y [res==0]
print(case0.value_counts(),'\n')  

case1 = Y[res==1]
print(case1.value_counts(),'\n')

case2 = Y[res==2]
print(case2.value_counts())
```

```python
## visualizacion de los centroides de cada grupo 
labels = {'Virginica','Versicolor','Setosa'}
np.round(pd.DataFrame(kmeans.cluster_centers_, columns = header_x , index = ['cluster 1 ','Cluster 2','Cluster3'] ),2)
```

```python
##grafico de barras cluster 0
sns.countplot(x = case0).set_title('Cluster 0 ')
plt.show()
```

```python
sns.countplot(x = case1).set_title('Cluster 0 ')
plt.show()
```

```python
sns.countplot(x = case2).set_title('Cluster 2 ')
plt.show()
```

```python
#Predicciones 
X_test = {'sepal_length':[7.0],'sepal_width':[3.0],'petal_length':[5.0],'petal_width':[1.5]}
X_test_df = pd.DataFrame(X_test)

pred = kmeans.predict(X_test_df)[0]
print('Prediccion en el cluster {} con la etiqueta {} '.format(pred,labels[pred]))
```
