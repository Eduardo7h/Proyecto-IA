```python
# header = 'infer' toma la primer fila como titulos  
df = pd.read_csv('data_studentlist.csv', header = 'infer')
df.head()
```

```python
#Modificar encabezado de df
df = df.rename(columns = {
    "name":"NAME",
    "age": "AGE"
})
df.head()
```

```python
# encabezado de columnas se pone en mayusculas 
df.columns = [ col.upper() for col in df.columns]
df.head()
```

```python
#en cabezado de columnas se pon en minisculas 
df.columns = [ col.lower() for col in df.columns ]
df.head()
```

```python
#agrega al inicio 
df = df.add_prefix("prefix_")
#agrega al final 
df = df.add_suffix("_suf")
df.head()
```

```python
df.columns = ['NAME', 'GENDER' , 'AGE', 'GRADE', 'ABSENCE', 'BLOODTYPE', 'HEIGHT', 'WEIGHT']
df.head()
```

```python
#Agregar una nueva columna a los datos 
df['IMC'] = 1000*df['WEIGHT']/df['HEIGHT']**2
df.head()
```

```python
df_left = df.loc[:,['NAME','AGE','GENDER','ABSENCE']]
df_right = df.loc[:,['NAME','BLOODTYPE','WEIGHT','HEIGHT']]

#union df_izq y df_dere , especificar la columna , se devuelve comom una interseccion 
pd.merge(df_left, df_right, left_on='NAME', right_on='NAME', how ="inner")
```

```python
df_left_peque = df_left.loc[:10,:]
df_right_peque =df_right.loc[7:,:]

#une las tables una arriba ade otra en orden alfabetico  
pd.concat([df_left_peque,df_right_peque], sort = True)

#axis nos dive sobre que eje opera 1:columnas , 0:filas 
# une dos tablas una a lado de la otra 
pd.concat([df_left_peque,df_right_peque], axis = 1,sort = True)
```

tabla pibote: forma de reorganizar una tabla ya existente o de ordenar otra tabla de datos (reordenar datos) 

```python
import numpy as np
import pandas as pd
import os 
```

```python
my_dict = {"Size": ["L", "L", "M", "M", "M", "S", "S", "S", "S"],
            "Type": ["A", "A", "A", "B", "B", "A", "A", "B", "B"],
            "Location": ["L1", "L1", "L1", "L2", "L2", "L1", "L2", "L2", "L1"],
            "A": [1, 2, 2, 3, 3, 4, 5, 6, 7],
            "B": [2, 4, 5, 5, 6, 6, 8, 9, 9]}
df =pd.DataFrame(my_dict)
df
```

```python
#creacion de tabla pibote 
dfr = pd.pivot_table(df,index = ['Size','Type'],columns = 'Location', values = 'B')
dfr

# las casillas que tienen NAN se van a reemplzar con 0 
dfr = pd.pivot_table(df,index = ['Size','Type'],columns = 'Location', values = 'B', fill_value = 0)
dfr
```

```python
dfr_1 = pd.pivot_table(df , index =['Size','Type'], columns = 'Location', values = 'B',aggfunc = np.median, fill_value = 0)
dfr_1
```

```python
pd.pivot_table(df, index=['Size','Type'], values = ['A','B'],
               aggfunc ={'A':np.max,'B' :np.min})
# funciones de agregacion aggfuc indica con que valores se van a llenar 

```

```python
#data frame con valores en NAN
datos = {
    'precio': [34, 54, np.nan, np.nan, 56, 12, 34],
    'cantidad_en_stock': [3, 6, 14, np.nan, 5, 2, 10],
    'productos_vendidos': [3, 45, 23, np.nan, 24, 6, np.nan]
}

df = pd.DataFrame(datos)
df.head()
```

```python
mascara = df.isna()
mascara

#cuantos valores nulos tiene cada columna 
df.isna().sum(axis=0)

#cuantos valores nulos tiene cada fila 
df.isna().sum(axis=1)
```

```python
#eliminar valores faltantes 
df.dropna(axis=0, how='any')#elimna la fila con almenos un valor nulo 
df.dropna(axis=0, how='all')#elimna la fila con mas de un valor null
df.dropna(axis=0,thresh=2)#thresh umbral conserva las filas que tienen al menos dos null
```

```python
#crear una nueva columna con NAN
df['descuento'] = np.nan
df

df.dropna(axis=1,how='any')
```

```python
df_sin_nan = df.dropna(axis=1,how='all') 
df_sin_nan = df_sin_nan.dropna(axis=0,how='all')
df_sin_nan
```

```python
#modificar los nan con 0 
df_sin_nan['productos_vendidos'] = df_sin_nan['productos_vendidos'].fillna(0)
#solo modifico los productos vendidos 
df_sin_nan

#modifica los NAN de precio 
df_sin_nan['precio'] = df_sin_nan['precio'].fillna(df_sin_nan['precio'].mean())
#solo modifico los NAN en precio 
df_sin_nan
```
