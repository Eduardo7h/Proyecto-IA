```python
import matplotlib.pyplot as plt 
import numpy as np

```

```python
x = np.array(['Q1','Q2','Q3','Q4'])
y = np.array([234.0 , 245.7 , 144.6 , 317.6])
plt.bar(x,y,color = 'aqua')
plt.title('Grafico de barras')
plt.show()
```
