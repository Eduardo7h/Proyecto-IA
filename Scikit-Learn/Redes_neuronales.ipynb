# Perceptron Multicapa 
resuelve el problema con la operacion logica XOR, lo que hace es que hay mas de un percentron activo al mismo tiempo 
- las entradas son seprables no linalmente   <br>

una solucion es transformar el espacio de un plano a un hiperplano, para poder doblarlo <br>
 


- una solucion para la compuerta logica XOR con los perceptron multicapa. Teniendo en cuenta que se va a sumar la llamada capa oculta (transforma el espacio), cuando tenenmos este tipo de transformacion se usa una funcion no lineal en este caso la sigmoide <br>



## Idea del perceptron multicapa 


- red neuronal - perceptrones en cadenados que transforma a un problema linealmente separable 

- Propagacion hacia adelante: a mis datos de entrada los multiplico por los pesos y le aplico una activacion para obtener una nueva entrada, asi susesiva mente hasta obtener la salida 

# Retropropagacion
 propagacion hacia atras, tomar el error relativo (y(original)-y(pred)) y vamos retrocediendo en las capas utilizando el descenso de gradiente (proceso de obtimizacion para actualizar los pesos)
<br>


tasa de aprendizaje - ayuda a controlar que tan grande voy a sia encontrar el minimo 

# Funciones de activacion

- Funcion lineal: se utiliza solo en la capa de salida <br>


- Funcion Tanh: transforma la suma ponderada de entradas antes de pasarla a la siguiente capa 


- ReLu: esta funcion evita que tengamos gradientes minimos, utilizada en redes profunda, esta funcion tambien puede presentear problemas con ciertas neuronas <br>


- Leawky ReLu: tiene una modificacion para evitar neuronas muertas <br>

Ejemplos de funciones de costo <br>


- Optimizacion: proceso de encontrar la mejor solucion posible a un problema, ya sea maximizando o minimizando una funcion
1. optimo local - mejor solucion en una vecindad determinada dentro de una region factible, pero puede no ser la mejor de todo el conjunto 
2. optimo global - mejor solucion de un conjunto de datos 

- Gradiente : vector de derivadas parciales de una funcion respecto a cada una de sus variabls (el gradiante a punta hacia la direccion del maximo)

- magnitud: taza de cambio maximo en sierta direccion 

- parametros del modelo - parametros internos que el modelo ajusta para capturar la relacion entre los valores de entrada y salida <br>



# Convexo VS no convexo

- Convexo: un solo valle -> minimo global 
- NO convexo: varios valles -> minimos locales + puntos silla 

# Descenso del gradiente 
metodo iterativo que se utiliza para minimizar la funcion de perdida
1. se define un punto de partida y se determina la taza de aprendizaje(parametro que nos ayuda a ver que tanto bajamos), toleracia hunbral que determina si el algoritmo ha terminado 

2. Calcular el gradiente de la funcion de perdida. Este vector indica la direccion en la que la funcion aumenta de forma mas rapida 

3. Actualizacion de los parametros. se actualizan los parametros moviendose en la direccion opuesta al gradiente(para minimizar L)

4. Repetir el proceso n veces hata cumplir ciertos criterios de convergencia <br>

# Resumen 
__Peso__ -> pondera la importancia de las enntradas asi como la interconexion entre las neuronas 

__funcion de activación__ -> decide si la neurona se activa o no y que tanto se activa 

__Sesgo__ -> Ajusta la activacion de la nuerona sin depender de las entradas  

__Hiperparametro__ -> son parametros que se ajustan previamente antes del entrenamiento 

__conjunto de entrenamiento__: este sirve para entrenar el modelo y definir los valores de los pesos 

__conjunto de validacion__: en esta parte se evaluan los pesos y se modifican en caso de ser necesario 

__conjunto de prueba__: sirve para evaluar tu modelo, son datos nunca hantes vistos 

__Funcion de Perdida__ -> nos ayuda a medir el error de prediccion de la red neuronal 

__Backpropagation__ -> es el proceso en el que el error de salida se propaga hacia atras a traves de la red, ajustandose en cada neurona segun su contribucion al resultado 

__Gradiente descendente__ -> es un metodo para optmizar el error, esto lo logra ajustando parametros del modelo  


__Epoca__ -> pasada completa de todos los datos de entramiento 

__Batch__ -> subconjuto de datos de entrenamiento usados para actualizar los pesos / (Iteraciones)
