## Aprendizaje no supervisado 
este tipo de aprendizaje aprende con daton sin etiquetar por lo que tiene la tarea de idintificar patrones y estadisticas sin ninguna orientacion o instruccion explicita. 

este tipo de aprendizaje utiliza las tecnicas de 
- Agrupacion: se refiere a hacer grupos de objetos o personas similares de acuerdo con el proposito del analis y el algotrimo 
- Asociacion: se refiere a descubrir una relacion de co-ocurrencia entre elementos especificos en el conjunto d datos   
- Reduccion de dimensionalidad: se refiera a resumir un conjunto de variables dadas en un numero mas pequeño de variables para proporcionar explicaciones efectivas.    

# Analisis de conglomerados 

la agrupacion en clustering es un algoritmo que organiza y clasifica diferentes objetos, puntos de datos u observaciones basados en similitudes o patrones, el agrupamiento se ocupa principalmete para conocer los atributos de diferentes grupos y darnos una idea de los patrones subyacentes de diferentes grupos en lugar de hacer una prediccion. 
  



- Agrupación en Clsutering jerarquico : Este método determina los clustering en funcion de lo cerca que están los puntos de datos entre si en todas las dimensiones. La idea es que los objetos que estan más cerca ésten más relacionados que los que están lejos entre si  

este tipo de agrupacion crea una red de garficos de los clústeres en cada nivel jerarquico. Esta red es jerarquica, lo que significa que cualquier nodo solo tiene un nodo principal, pero puede tener varios nodos secundarios  <br>



- Cluster no jerarquicos: este tipo de metodo clasifican los elementos segun un numero de cluster dado, necesitan que el numero de particiones esten fijados apriori

# Método jerarquico aglomerativo 
comienza con puntos de datos individuales y fuciona sucesivamente los clusters calculando la matriz de proximidad (normalmente usando la similitud coseno) de todos los clusteres en el nivel actual de la jerarquia para crear estructuras similares a un arbol 



```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt 

from sklearn.datasets import make_blobs, make_moons
from sklearn.cluster import AgglomerativeClustering
```

```python
X1, label1 = make_blobs(n_samples=200, n_features=2, centers=2, cluster_std=5, random_state=123)
plt.scatter(X1[:,0],X1[:,1], c=label1, alpha=0.7)
plt.title('Dataset #1 : Original')
plt.show()
```

```python
X2, label2 = make_moons(n_samples=200,noise=0.08,random_state=123)
plt.scatter(X2[:,0],X2[:,1], c=label2, alpha=0.7)
plt.title('Dataset #1 : Original')
plt.show()
```

```python
## Dos Clusteres 
agglo = AgglomerativeClustering(n_clusters=2)
agglo.fit(X1)
my_colors = {0:'red',1:'green'}
plt.scatter(X1[:,0],X1[:,1], c= pd.Series(agglo.labels_).apply(lambda x: my_colors[x]),alpha=0.7)
plt.title('Dataset 1')
plt.show()
```

```python
## Dos Clusteres 
agglo = AgglomerativeClustering(n_clusters=2)
agglo.fit(X2)
my_colors = {0:'red',1:'green'}
plt.scatter(X2[:,0],X2[:,1], c= pd.Series(agglo.labels_).apply(lambda x: my_colors[x]),alpha=0.7)
plt.title('Dataset 2')
plt.show()
```

## Determinar K

El método más utilizado para determinar el número óptimo de clusters es el método del codo. Este consiste en entrenar el modelo con diferentes valores de 𝑘 y calcular qué tan similares son los datos dentro de cada conglomerado mediante la suma de los cuadrados intra-cluster (WCSS).

Posteriormente, se construye una gráfica en la que se observa cómo cambia esta medida en función de 𝑘. El punto donde la gráfica presenta una inflexión (forma de “codo”) se considera el valor óptimo de k .
