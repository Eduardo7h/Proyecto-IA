```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 

from sklearn.datasets import make_blobs, make_moons
from sklearn.cluster import KMeans,DBSCAN,AgglomerativeClustering
```

```python
X, label = make_blobs(n_samples=300,n_features=2,centers=3,cluster_std=5,random_state=123)
plt.scatter(X[:,0],X[:,1], c=label, alpha=0.7)
plt.title('Dataset #1 Original')
plt.show()

```

```python
## Generamos el modelo K-means 

KMeans = KMeans(n_clusters=3).fit(X)

#asignamos un determinado cluster 
KMeans.labels_

print('Posicion de los centroides: \n', KMeans.cluster_centers_) # arreglo de numpy con la posicion de cada centroide  

plt.figure(figsize=(6,5))

## Creamos la grafica 
for cluster in range(KMeans.n_clusters):
    plt.scatter(X[:,0], X[:,1], c=KMeans.labels_, alpha=0.2)

    plt.scatter(
        KMeans.cluster_centers_[cluster][0],
        KMeans.cluster_centers_[cluster][1],
        marker='P',s=200,color='black')
    
plt.title('K-means')
plt.show()

```

```python
#epsilon - radio del circulo de vecindad alrededor de un punto 
# min_samples - minimo de puntos dentro del circulo 
cluster = DBSCAN(eps=2,min_samples=5).fit_predict(X)

# Crear el diagrama de dispersión
plt.figure(figsize=(6, 5))

plt.scatter(X[:, 0], X[:, 1],
            c=cluster, s=50, alpha=0.7)

plt.title("DBSCAN")
plt.show()

```

```python
## Agrupamiento jerarquico 
jerarquico = AgglomerativeClustering(n_clusters=3).fit(X) 
jerarquico.labels_

## Creamos el diagrama de dispersión 
plt.figure(figsize=(6,5))


plt.scatter(X[:, 0], X[:, 1],
            c=jerarquico.labels_, cmap='viridis', s=50,alpha=0.7)

    
plt.title('Agrupamiento jerarquico')
plt.show()


```
