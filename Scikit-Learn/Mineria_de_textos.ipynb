```python
my_text = """Machine learning is the scientific study of algorithms
and statistical models that computer systems use to effectively perform
a specific task without using explicit instructions, relying on patterns 
and inference instead. It is seen as a subset of artificial intelligence. 
Machine learning algorithms build a mathematical model of sample data, known 
as "training data", in order to make predictions or decisions without being 
explicitly programmed to perform the task. Machine learning algorithms are used
in the applications of email filtering, detection of network intruders, and computer 
vision, where it is infeasible to develop an algorithm of specific instructions 
for performing the task. Machine learning is closely related to computational statistics,
which focuses on making predictions using computers.The study of mathematical optimization 
delivers methods, theory and application domains to the field of machine learning.
Data mining is a field of study within machine learning, and focuses on exploratory data 
analysis through unsupervised learning In its application across business problems, machine 
learning is also referred to as predictive analytics."""

my_text = [my_text.lower()]    ## creamos una lista con el texto en minusculas 
```

```python
## n-gramas 
n = 3
n_gram_type = 'word'   ##palabras 
Vectorizer = CountVectorizer(ngram_range=(n,n),analyzer=n_gram_type)

n_grams = Vectorizer.fit(my_text).get_feature_names_out()  

## convertimos el texto a una matriz numerica
n_grams_cts = Vectorizer.transform(my_text).toarray() 
n_grams_cts = list(n_grams_cts[0])
```

```python
## Emparejar el conteo con n-gramas 
list(zip(n_grams,n_grams_cts))

## declaramos un diccionario 
mi_dic= {}
for a_gram in n_grams:
    words = nltk.word_tokenize(a_gram)  ## lista de palabras separadas 
    a_nm_gram = ' '.join(words[0:n-1]) #unimos las palabras con espacios (contexto)
    next_word = words[-1]
    ## si el contexto no existe en el diccionario
    if a_nm_gram not in mi_dic:
        mi_dic[a_nm_gram] = [next_word]  # si no existe el contexto se crea 
    else:
        mi_dic[a_nm_gram] += [next_word]  # si ya existe se añade la sigue palabra 
```

```python
## funcion que predice la siguiente palabra dada las dos anteriores 
def predict_next(a_nm_gram):
    value_list_size = len(mi_dic[a_nm_gram]) #regrasa cuantas opciones hay para el contexto 
    i_pick = randint(0,value_list_size) #elige un numero al azar entre 0 y la longitud de las opciones del contexto 
    return (mi_dic[a_nm_gram][i_pick])

```

```python
# Prueba de la funcion 
input_str ='machine learning'
print(predict_next(input_str)) 

for i in range(10):
    print(predict_next(input_str))
```

```python
import numpy as np 
import re
import seaborn as sns 
import matplotlib.pyplot as plt 

from nltk.corpus import stopwords
from sklearn.datasets import load_files
from sklearn.model_selection import train_test_split 
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score 
```

```python
reviews = load_files('txt_sentoken')  ## leemos los archivos 
my_docs , y = reviews.data, reviews.target  ## coloca en my_doc los valores ,y son etiquetas si es 0 o 1
print(reviews.target_names)
```

```python
sns.countplot(x=y).set_title('Distribution of clases')
plt.show()
```

```python
## Preprocesamiento 
corpus =[]  ## lita vacia para guardar los archivos limpios 
for i in range(0,len(my_docs)):  ## leemos cada review 
    reviews = re.sub(r'\W', ' ',str(my_docs[i]))   #eliminamos signos de puntuacion y los transforma en espacios 
    reviews = reviews.lower()    ##Pasamos todo a minusculas 
    reviews = re.sub(r'^br$',' ', reviews)       #^inicio, $ final de parafo, reemplazamos por espacio 
    reviews = re.sub(r'\s+br\s',' ',reviews)
    # eliminamos palabras de una sola letra 
    reviews = re.sub(r'\s+[a-z]\s+',' ',reviews)
    reviews = re.sub(r'^b\s+','',reviews) 
    # Removemos espacios entre oraciones 
    reviews = re.sub(r'\s+',' ',reviews)
    corpus.append(reviews)
```

```python
TFIDF = TfidfVectorizer(max_features=1000,min_df=2,max_df=0.6,
                        stop_words = stopwords.words('english'))   # elegimos las 1000 palabras mas frecuentes 
TFIDF.fit(corpus)
X = TFIDF.transform(corpus)      ## de texto a matriz
corpus.append(reviews) 
X.shape
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=1234)
```

```python
LR = LogisticRegression()
LR.fit(X_train,y_train)
```

```python
y_pred = LR.predict(X_test)
conf_ma = confusion_matrix(y_test,y_pred)
acc = accuracy_score(y_test,y_pred)
```

```python
print(conf_ma)
print(f'Accuracy:{acc:.3f}')
```

```python
def limpiar(text):
    text = str(text)
    text = text.lower()
    text = re.sub(r'^br$',' ', text)       #^inicio, $ final de parafo, reemplazamos por espacio 
    text = re.sub(r'\s+br\s',' ',text)
    # eliminamos palabras de una sola letra 
    text = re.sub(r'\s+[a-z]\s+',' ',text)
    text = re.sub(r'^b\s+','',text) 
    # Removemos espacios entre oraciones 
    text = re.sub(r'\s+',' ',text)
    return text.strip()
```

```python
def predic_review(text):
    tex_clean = limpiar(text)
    vec = TFIDF.transform([text_clean])
    pred = LR.predict(vec)[0]
    
    return reviews.target_names[pred]
```

```python
my_doc = [
'AI is a rapidly advancing field that involves the development of intelligent machines',
'Machine learning is a subset of AI that focuses on training machines to learn from data,',
'Deep learning is a subfield of machine learning that uses multi-layered neural networks,',
'AI is applied in various sectors, such as healthcare, transportation, and entertainment.',
'Ethical considerations play an important role in the development of AI.',
'Artificial intelligence can revolutionize many aspects of society.',
'AI systems, such as chatbots and virtual assistants, are becoming increasingly common.',
'Natural language processing is a branch of AI that enables machines to understand human language.',
'The goal of computer vision is for machines to understand visual information'
]
```

```python
## Declaramos el estimador 
vectorizer = TfidfVectorizer(max_features=18,min_df=1,max_df=3,
                        stop_words=stopwords.words('english'))
X = vectorizer.fit_transform(my_doc)   ## Vectores numerico de cada doc 
```

```python
matriz_cos = cosine_similarity(X)   ## obtenemos la matriz de similitud coseno 
matriz_cos
```

```python
np.fill_diagonal(matriz_cos,0)  ## ignoramos la diagonal principal de la matriz 
promedio_simi = matriz_cos.mean(axis=1)   ## promedio de las filas 
valor_mayor = np.argmax(promedio_simi)    ## guarda el mayor promedio
print('Documento mas similar a  los demas  -> ',valor_mayor) 
```
