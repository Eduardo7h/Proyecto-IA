## AutoEncoder 
bajan la dimensionalidad de las imagenes/Datos poniendo varias capas ocultas para ir reduciondo la dimensionalidad 
comprime y descomprime datos en la fase del entrenamiento.

genera informacion apartir de informacion ya existentes 



<br>
el AutoEncoder se pude dividir en doas partes el codificador y el decodificador, se pueden diseñar autoencoder 
con diversos 

tiempo de entrenamiento  se reduce si bajamos la dimensionalidad de los datos 

## Aplicaciones 
1. reduccion de la dimensionalidad 
2. compresión de inagenes 
3. para quitar ruido de imagenes o restaurarlas 
4. Generacion de imagenes
5. Deepfake 
6. sistemas de recomendacion(con base en gustos y referencias)

## Redes GAN
la arquitectura GAN implica dos sub-modelos: un modelo generado para generar nuevos ejemplos. se consideran adversarios 

- modelo generador: el modelo generado toma un vector aleatorio de longuitud fija como entrada y genera una muestra.genera un lote de muestras 

- modelo discriminador: toma un ejemplo del dominio como entrada (real o generada) y predice una etiqueta de clase binario. actualiza para mejorar la discriminacion de muetsras reales y falsas
